```python
import pandas as pd
import nltk
nltk.download('punkt_tab')
import textstat
from utils import MetricHelper
import tqdm

root_cause_c_df = pd.read_csv('Root_Cause_Christian.csv')
root_cause_d_df = pd.read_csv('Root_Cause_David.csv').rename(columns={"hit": "bug", "gold_standard": "description"})

root_cause_c_dict = root_cause_c_df.set_index('bug')['description'].to_dict()
root_cause_d_dict = root_cause_d_df.set_index('bug')['description'].to_dict()

metrics = {}
for key, value in tqdm.tqdm(root_cause_c_dict.items()):
    m = {}
    m['rouge'] = MetricHelper.calculateRougeScore(value, root_cause_d_dict[key])
    m['bleu'] = MetricHelper.calculateBleuScore(value, root_cause_d_dict[key])
    # m['fk'] = MetricHelper.calculateFKReadability(value)
    m['fk'] = MetricHelper.calculateFKScore(value)
    m['bleurt'] = MetricHelper.calculateBleurtScore([value], [root_cause_d_dict[key]])
    metrics[key] = m

metrics
```
